```python
import pandas as pd
import os

# Folders containing .psv files
folders = ["../data/raw/training_setA", "../data/raw/training_setB"]

def read_all_psv(folders):
    """
    Reads all .psv files from a list of folders and combines them into one DataFrame.
    """
    df_list = []
    for folder_path in folders:
        # Loop over all .psv files in the folder
        for file_name in os.listdir(folder_path):
            if file_name.endswith(".psv"):
                file_path = os.path.join(folder_path, file_name)
                df = pd.read_csv(file_path, sep='|')
                df_list.append(df)
    
    # Combine all DataFrames into one
    combined_df = pd.concat(df_list, ignore_index=True)
    return combined_df

# Read and combine all .psv files from both folders
train = read_all_psv(folders)

# Quick check
print(train.shape)
print(train.head())
```
